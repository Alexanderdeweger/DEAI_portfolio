```python
import itertools
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.datasets import fashion_mnist

results = []

# Define parameter grid
param_grid = {
    'layers': [[64, 32], [128, 64], [256, 128, 64], [32]],
    'epochs': [5, 10],
    'learning_rate': [0.001, 0.01],
    'batch_size': [32, 64]
}

(train_data, train_labels), (test_data, test_labels) = fashion_mnist.load_data()
train_data_norm = train_data / train_data.max()
test_data_norm = test_data / test_data.max()

# Generate all combinations
keys, values = zip(*param_grid.items())
combinations = [dict(zip(keys, v)) for v in itertools.product(*values)]

for params in combinations:
    model = Sequential()
    model.add(Flatten(input_shape=(28, 28)))
    for nodes in params['layers']:
        model.add(Dense(nodes, activation='relu'))
    model.add(Dense(10, activation='softmax'))

    model.compile(
        loss='sparse_categorical_crossentropy',
        optimizer=Adam(learning_rate=params['learning_rate']),
        metrics=['accuracy']
    )

    history = model.fit(
        train_data_norm, train_labels,
        epochs=params['epochs'],
        batch_size=params['batch_size'],
        validation_data=(test_data_norm, test_labels),
        verbose=0
    )

    val_acc = max(history.history['val_accuracy'])
    results.append({**params, 'val_accuracy': val_acc, 'layers': str(params['layers'])})
    print(f"Done: {params} → {val_acc:.4f}")

df = pd.DataFrame(results)
df
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for ax, param in zip(axes, ['learning_rate', 'epochs', 'batch_size']):
    grouped = df.groupby(param)['val_accuracy'].mean()
    grouped.plot(kind='bar', ax=ax, title=f'Avg Val Accuracy by {param}')
    ax.set_ylabel('Validation Accuracy')
    ax.set_ylim(0.7, 1.0)

plt.tight_layout()
plt.savefig("experiment_results.png")
plt.show()
```
